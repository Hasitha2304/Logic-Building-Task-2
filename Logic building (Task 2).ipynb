```python
# 1. Unique words in a sentence
sentence = input("Enter sentence: ")
words = sentence.split()
unique_words = []
for w in words:
    if w not in unique_words:
        unique_words.append(w)
print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)


# 2. Highest salary from employee data
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}
highest_salary = 0
highest_employee = ""
for name in employees:
    if employees[name] > highest_salary:
        highest_salary = employees[name]
        highest_employee = name
print("Highest Salary:", highest_employee, "-", highest_salary)


# 3. Find maximum and minimum values
numbers = [45, 22, 89, 10, 66]
max_num = numbers[0]
min_num = numbers[0]
for n in numbers:
    if n > max_num:
        max_num = n
    if n < min_num:
        min_num = n
print("List:", numbers)
print("Max:", max_num)
print("Min:", min_num)


# 4. Count products above a price threshold
prices = [450, 1200, 899, 1500, 300]
count = 0
for p in prices:
    if p > 1000:
        count += 1
print("Products above 1000:", count)


# 5. Calculate attendance percentage
attendance = ["P", "P", "A", "P", "P"]
present_count = 0
for a in attendance:
    if a == "P":
        present_count += 1
percentage = (present_count / len(attendance)) * 100
print("Attendance Percentage:", percentage)


# 6. Remove duplicate phone numbers
phone_numbers = [9876543210, 9123456789, 9876543210]
unique_numbers = set(phone_numbers)
print("Unique phone numbers:", unique_numbers)


# 7. Count character frequency
text = input("Enter a string: ")
freq = {}
for ch in text:
    if ch in freq:
        freq[ch] = freq[ch] + 1
    else:
        freq[ch] = 1
print(freq)


# 8. Convert list to tuple
lst = [10, 20, 30]
tup = tuple(lst)
print("Tuple:", tup)


# 9. Check if a key exists in a dictionary
employee_data = {"Ravi": 75000, "Anita": 68000, "Kiran": 72000}
key_to_check = input("Enter employee name to check: ")
if key_to_check in employee_data:
    print("Employee exists")
else:
    print("Employee does not exist")


# 10. Calculate average marks
marks = [70, 85, 60, 75]
total = 0
for m in marks:
    total = total + m
average = total / len(marks)
print("Average Marks:", average)
```
